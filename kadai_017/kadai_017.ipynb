```python
class Human:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def check_adult(self):
        if self.age >= 20:
            print("大人")
        else:
            print("大人ではない")

menbers = {"太郎": 33, "二郎": 30, "三郎": 24, "四郎": 19, "五郎": 16}

for menber in menbers.items():
    print(Human(menber[0], menber[1]).check_adult())
```
