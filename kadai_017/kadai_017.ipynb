```python
class Human:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def check_adult(self):
        if self.age >= 20:
            print("大人")
        else:
            print("大人ではない")

taro = Human("太郎", 33)
jiro = Human("二郎", 30)
saburo = Human("三郎", 19)

menbers_list = [taro, jiro, saburo]
for menber in menbers_list:
    menber.check_adult()
```
